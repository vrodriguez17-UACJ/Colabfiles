## Parte 1 — Vectores: perfiles de producto

Una tienda en línea representa cada producto como un vector de 3 componentes: calificación de calidad, calificación de precio, y calificación de servicio (escala 1 a 5, donde 5 es mejor). Ejecuta la siguiente celda para cargar los tres productos — **no la modifiques**.

````python
producto_A = np.array([4.5, 3.2, 5.0])
producto_B = np.array([4.0, 2.8, 4.7])
producto_C = np.array([1.0, 5.0, 1.2])

print(producto_A, producto_B, producto_C)
````

---

### Ejercicio 1 — Norma de los vectores (10 pts)

Calcula la norma de `producto_A`, `producto_B` y `producto_C` usando `np.linalg.norm()`.

````python
# Tu código aquí
norma_A = np.linalg.norm(producto_A)
norma_B = np.linalg.norm(producto_B)
norma_C = np.linalg.norm(producto_C)

print(f"Norma producto_A: {norma_A:.4f}")
print(f"Norma producto_B: {norma_B:.4f}")
print(f"Norma producto_C: {norma_C:.4f}")
````

---

### Ejercicio 2 — Producto punto y ángulo (15 pts)

1. Calcula el producto punto entre `producto_A` y `producto_B`, y entre `producto_A` y `producto_C`, usando `np.dot()` o `@`.
2. Calcula el ángulo (en grados) entre `producto_A` y `producto_B`, y entre `producto_A` y `producto_C`.
3. Con base en los ángulos obtenidos, indica (en un comentario o celda de texto) cuál de los dos productos (`B` o `C`) es más similar a `producto_A`, y por qué el ángulo es el indicador correcto para esa comparación.

````python
# Tu código aquí
# 1. Producto punto
dot_AB = np.dot(producto_A, producto_B)
dot_AC = np.dot(producto_A, producto_C)
print(f"1. Producto punto (A · B): {dot_AB:.4f}")
print(f"   Producto punto (A · C): {dot_AC:.4f}")

# 2. Ángulo en grados entre vectores
cos_AB = dot_AB / (norma_A * norma_B)
cos_AC = dot_AC / (norma_A * norma_C)

angulo_AB = np.degrees(np.arccos(cos_AB))
angulo_AC = np.degrees(np.arccos(cos_AC))
print(f"\n2. Ángulo entre A y B: {angulo_AB:.4f}°")
print(f"   Ángulo entre A y C: {angulo_AC:.4f}°")
````

Comparación por ángulo: El producto_B es mucho más similar al producto_A, ya que el ángulo entre ambos es de apenas $1.77°$ (casi colineales), frente a los $47.23°$ que separan a producto_A y producto_C. El ángulo es el indicador adecuado porque mide la orientación o proporción relativa entre los atributos (calidad, precio, servicio) independientemente de la magnitud de las calificaciones: A y B mantienen el mismo perfil proporcional de evaluación, mientras que C apunta en una dirección distinta (puntaje alto en precio pero muy bajo en calidad y servicio).

### Ejercicio 3 — Normalización (10 pts)

Normaliza los tres vectores (`producto_A`, `producto_B`, `producto_C`) para obtener sus vectores unitarios. Verifica, imprimiendo el resultado, que la norma de cada vector normalizado sea 1.

````python
# Tu código aquí
# Obtener vectores unitarios dividiendo cada vector entre su norma
unit_A = producto_A / norma_A
unit_B = producto_B / norma_B
unit_C = producto_C / norma_C

print("Vector unitario A:", np.round(unit_A, 4), "| Norma:", np.linalg.norm(unit_A))
print("Vector unitario B:", np.round(unit_B, 4), "| Norma:", np.linalg.norm(unit_B))
print("Vector unitario C:", np.round(unit_C, 4), "| Norma:", np.linalg.norm(unit_C))
````

## Parte 2 — Matrices: mezcla de insumos

Una fábrica de jugos combina dos insumos (concentrado de fruta y agua) en distintas proporciones para dos líneas de producto. La matriz `M` representa esas proporciones (filas = línea de producto, columnas = insumo), y `P` representa un ajuste de escala para una nueva presentación. Ejecuta la siguiente celda — **no la modifiques**.

````python
M = np.array([[2, 1],
              [1, 3]])

P = np.array([[1.5, 0],
              [0, 2.0]])

print(M)
print(P)
````

---

### Ejercicio 4 — Multiplicación matricial (15 pts)

1. Calcula `M * P` (elemento a elemento) y `M @ P` (multiplicación matricial). Los resultados son distintos — en un comentario o celda de texto, explica brevemente por qué.
2. Verifica que las formas sean compatibles: ¿cuál sería el resultado de `P @ M`? Calcúlalo y compáralo con `M @ P` — ¿da el mismo resultado?

````python
# Tu código aquí
# 1. Multiplicación elemento a elemento vs multiplicación matricial
elem_a_elem = M * P
mat_MP = M @ P

print("1. M * P (elemento a elemento):\n", elem_a_elem)
print("\n   M @ P (multiplicación matricial):\n", mat_MP)

# 2. Compatibilidad de formas y cálculo de P @ M
print(f"\n2. Formas -> M: {M.shape}, P: {P.shape} (compatibles para P @ M)")
mat_PM = P @ M
print("   P @ M:\n", mat_PM)
print("   ¿M @ P es igual a P @ M?:", np.array_equal(mat_MP, mat_PM))
````

¿Por qué M * P y M @ P dan resultados distintos? Porque M * P realiza el producto de Hadamard (elemento a elemento) multiplicando únicamente posiciones homólogas ($M_{ij} \cdot P_{ij}$, anulando las posiciones fuera de la diagonal porque $P$ tiene ceros allí), mientras que M @ P realiza el producto matricial (combinación lineal de filas de $M$ por columnas de $P$, escalando cada columna de $M$ por el factor correspondiente de $P$).¿P @ M da el mismo resultado que M @ P? No, porque la multiplicación de matrices no es conmutativa ($MP \neq PM$). Mientras M @ P escala las columnas de M (por $1.5$ y $2.0$), P @ M escala las filas de M por esos mismos factores.

### Ejercicio 5 — Determinante e inversa (15 pts)

1. Calcula el determinante de `M` con `np.linalg.det()`. Confirma que `M` sí tiene inversa (determinante distinto de 0).
2. Calcula la inversa de `M` con `np.linalg.inv()`.
3. Verifica tu resultado calculando `M @ M_inv` — debe dar (aproximadamente) la matriz identidad.

````python
# Tu código aquí
# 1. Determinante de M
det_M = np.linalg.det(M)
print(f"1. Determinante de M: {det_M:.4f} (distinto de 0 -> sí tiene inversa)")

# 2. Inversa de M
M_inv = np.linalg.inv(M)
print("\n2. Inversa de M:\n", M_inv)

# 3. Verificación M @ M_inv ≈ Identidad
identidad_verif = M @ M_inv
print("\n3. Verificación M @ M_inv:\n", np.round(identidad_verif, 6))
````

### Ejercicio 7 — Síntesis: distancia y similitud (20 pts)

Retomando los tres productos de la Parte 1:

1. Calcula la **distancia euclidiana** entre `producto_A` y `producto_B`, y entre `producto_A` y `producto_C`, usando la norma de la diferencia entre los vectores (`np.linalg.norm(a - b)`).
2. Calcula la **similitud coseno** entre los mismos pares — es el cociente que ya calculaste en el Ejercicio 2 antes de aplicar `arccos()`: $\frac{a \cdot b}{\|a\| \|b\|}$.
3. Con base en ambas métricas (distancia y similitud coseno), concluye en una cadena de texto impresa cuál producto (`B` o `C`) es más parecido a `producto_A`, y si ambas métricas coinciden en la misma conclusión.

````python
# Tu código aquí
# 1. Distancia euclidiana
dist_AB = np.linalg.norm(producto_A - producto_B)
dist_AC = np.linalg.norm(producto_A - producto_C)
print(f"1. Distancia euclidiana (A, B): {dist_AB:.4f}")
print(f"   Distancia euclidiana (A, C): {dist_AC:.4f}")

# 2. Similitud coseno
sim_cos_AB = np.dot(producto_A, producto_B) / (np.linalg.norm(producto_A) * np.linalg.norm(producto_B))
sim_cos_AC = np.dot(producto_A, producto_C) / (np.linalg.norm(producto_A) * np.linalg.norm(producto_C))
print(f"\n2. Similitud coseno (A, B): {sim_cos_AB:.4f}")
print(f"   Similitud coseno (A, C): {sim_cos_AC:.4f}")

# 3. Conclusión impresa
print(
    f"\n3. Conclusión: El producto B es mucho más parecido al producto A que el producto C. "
    f"Ambas métricas coinciden plenamente en esta conclusión: entre A y B la distancia euclidiana es menor "
    f"({dist_AB:.4f} vs {dist_AC:.4f}) y la similitud coseno es prácticamente 1 ({sim_cos_AB:.4f} vs {sim_cos_AC:.4f})."
)
````
